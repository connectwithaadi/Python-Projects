```python
USERNAME = "admin"
PASSWORD = "admin@123"

MAX_ATTEMPTS = 5

attempt  = 1

while attempt <= MAX_ATTEMPTS:

  print("=" * 40)
  print(f"Login Attempt {attempt}/{MAX_ATTEMPTS}")
  print("=" * 40)

  try:
    username = input("Enter Username: ").strip()
    password = input("Enter Password: ").strip()

    if username == "":
      raise ValueError("Username khali nhi hona chahiye")

    if password == "":
      raise ValueError("Password khali nhi hona chahiye")

    if username != USERNAME:
      raise Exception("Incorrect username")

    if password != PASSWORD:
      raise Exception("Incorrect password")

  except ValueError as error:
    print(f"Input Error: {error}")

  except Exception as error:
    print(f"Login Failed: {error}")

  else:
    print("\nLogin Successful ho gya!")
    print("Humare system me aapka swagat hai")
    break

  finally:
    print("End of login attempt.\n")

  attempt +=1
if attempt > MAX_ATTEMPTS:
  print("Maximum login ki attempt ko pura kr liya tumne...")
```
