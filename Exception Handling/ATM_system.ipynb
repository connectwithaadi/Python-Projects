# ATM System

```python
 balance = 10000.0

def check_balance():
  print(f"\nAvailable Balance : {balance:.2f}")

def deposit():
  global balance

  try:
    amount = float(input("Enter deposit amount: "))

    if amount <=0:
      raise ValueError("Deposit amount must be greater than zero")

  except ValueError as error:
    print(f"Error: {error}")

  else:
    balance += amount
    print(f"\n{amount:.2f} deposited successfully")
    print(f"Updated Balance : {balance:.2f}")

  finally:
    print("Deposit transaction completed.\n")

def withdraw():
  global balance

  try:
    amount = float(input("Enter withdrawl amount: "))

    if amount <=0:
      raise ValueError("Withdrawl amount must be greater than zero")

    if amount > balance:
      raise ValueError("Insufficient balance")

  except ValueError as error:
    print(f"Error: {error}")

  except Exeption as error:
    print(f"Error: {error}")

  else:
    balance -= amount
    print(f"\n{amount:.2f} withdrawn successfully")
    print(f"Updated Balance : {balance:.2f}")

  finally:
    print("Withdrawl transaction completed.\n")

while True:

    print("=" * 40)
    print("ATM MENU")
    print("=" * 40)
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 1:
            check_balance()

        elif choice == 2:
            deposit()

        elif choice == 3:
            withdraw()

        elif choice == 4:
            print("\nThank you for using our ATM.")
            break

        else:
            print("Invalid menu choice.")

    except ValueError:
        print("Please enter a valid number.")

    finally:
        print("-" * 40)
```
