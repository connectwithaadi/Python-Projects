```python
password = input("Enter your password: ")

has_upper = False
has_lower = False
has_digit = False
has_special = False

special_characters = "!@#$%^&*()-_=+[]{};:'\",.<>?/\|`~"

for char in password:
  if char.isupper():
    has_upper = True
  elif char.islower():
    has_lower = True
  elif char.isdigit():
    has_digit = True
  elif char in special_characters:
    has_special = True

print("\n Password Analysis")
print("-" * 35)

if len(password) >=8:
  print("✓ Maximun Length")
else:
  print("Minimum Length (8 characters required)")


if has_upper:
  print("✓ Uppercase Present")
else:
  print("✗ Uppercase Missing")

if has_lower:
  print("✓ Lowercase Present")
else:
  print("✗ Lowercase Missing")

if has_special:
  print("✓ Special Character Present")
else:
  print("✗ Special Character Missing")

if has_digit:
  print("✓ Digit Present")
else:
  print("✗ Digit Missing")

if len(password) >= 8 and has_upper and has_lower and has_digit and has_special:
    print("\nPassword Strength : Strong")
elif len(password) >= 6:
    print("\nPassword Strength : Medium")
else:
    print("\nPassword Strength : Weak")
```
