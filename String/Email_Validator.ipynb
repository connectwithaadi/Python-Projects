```python
email = input("Enter your email: ").strip()

valid = True

if email.count("@") !=1:
  valid = False

elif email.startswith("@") or email.endswith("@"):
  valid = False

elif "." not in email.split("@")[1]:
  valid = False

elif " " in email:
  valid = False

elif email.endswith("."):
  valid = False

print("Validation Result of your Email")

if valid:
  print("Your email id is valid")
else:
  print("Your email id is invalid")
```
