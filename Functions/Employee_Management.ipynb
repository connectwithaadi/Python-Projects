# Employee Management System


```python
employees = []

def add_employee():
  emp_id = input("Employee ID: ")
  name = input("Employee Name: ")
  salary = float(input("Salary: "))

  employees.append({
      "id": emp_id,
      "name": name,
      "salary": salary
  })

  print("Employee Added Successfully ")


def view_employee():
  if not employees:
    print("No Employee Records. ")
    return

  print("\nEmployee Records")
  print("-" * 40)

  for employee in employees:
    print(
        f"ID: {employee['id']} |"
        f"Name: {employee['name']} |"
        f"Salary: {employee['salary']: 2f}"
    )

def search_employee():
  emp_id = input("Enter Employee ID: ")

  for employee in employees:
    if employee["id"] == emp_id:
      print(employee)
      return
  print("Employee Not found. ")

def update_employee():

  emp_id = input("Enter Employee ID: ")

  for employee in employees:

    if employee["id"] == emp_id:
      employee["salary"] = float(input("New Salary: "))
      print("Salary Updated successfully")
      return

  print("Employee not Found. ")


def delete_employee():

  emp_id = input("Employee ID: ")

  for employee in employees:

    if employee["id"] == emp_id:
      employees.remove(employee)
      print("Employee Deleted Successfully. ")
      return

  print("Employee not found")


def menu():
  print("\n -----Employee Management Krne wala system ----")
  print("1. Add Employee")
  print("2. View Employee")
  print("3. Search Employee")
  print("4. Update Employee")
  print("5. Delete Employee")
  print("6. Exit")


while True:
  menu()

  choice = input("Enter aapki mand pasand choice: ")

  if choice == "1":
    add_employee()

  elif choice == "2":
    view_employee()

  elif choice == "3":
    search_employee()

  elif choice == "4":
    update_employee()

  elif choice == "5":
    delete_employee()

  elif choice == "6":
    print("Chalo niklo")
    break

  else:
    print("invalid option choose kr liya ab tumne, sahi option select krona bhai")

```
