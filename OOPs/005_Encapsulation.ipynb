```python
class Bank:
  def __init__(self, name: str, balance: int):
    #Private attribute
    self.__balance: int = balance

  # Getter
  def get_balance(self):
    print(f"Current balance = {self.__balance}")

  # Setter
  def set_balance(self, new_amount):
    self.__balance = new_amount

  def deposite(self, amount: int):
    self.__balance += amount
    print(f"Amount deosited, current balance = {self.__balance}\n")

  def withdraw(self, amount):
    if amount > self.__balance:
      print("Not enough money in bank\n")

    else:
      self.__balance -= amount
      print(f"Amount withdrawn, current balance = {self.__balance}\n")
```

```python
acc = Bank("Aaadi",1999)
acc.deposite(1000)
acc.get_balance()
acc.withdraw(499)
```
