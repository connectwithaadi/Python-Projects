```python
dog = Dog()
dog.bark()
dog.eat()
dog.sleep()
```

```python
class Animal:
  def __init__(self, name:str, age: int):
    self.name = name
    self.age = age

  def eat(self):
    print("I am eating")

  def sleep(self):
    print("I am sleeping")

class Dog(Animal):
  def bark(self):
    print("I am barking")

  def display(self):
    print(f"Age is {self.age}")
```

```python
dog = Dog("Tonny",10)
dog.bark()
dog.display()
```

```python
class Animal:
  def __init__(self, name: str, age: int):
    print("Animal INIT")
    self.name = name
    self.age = age

  def eat(self):
    print("I am eating")

  def sleep(self):
    print("I am sleeping")

class Dog(Animal):
  def __init__(self, name: str, age: int, breed: str):
    super().__init__(name, age)
    self.breed = breed
    print("This is dog init")

  def bark(self):
    print("I am barking")

  def display(self):
    print(f"Age is {self.age}")
```

```python
dog = Dog("tonny",4,"jocky")
dog.display()
```
