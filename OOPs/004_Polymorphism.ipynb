override concept comes here

```python
class Animal:
  def __init__(self, name: str, age: int):
    print("Animal INIT")
    self.name = name
    self.age = age

  def eat(self):
    print("I am eating")

  def sleep(self):
    print("I am sleeping")

  def move(self):
    print("I am moving")

class Dog(Animal):
  def __init__(self, name: str, age: int, breed: str):
    super().__init__(name, age)
    self.breed = breed
    print("This is dog init")

  def bark(self):
    print("I am barking")

  def display(self):
    print(f"Age is {self.age}")

  def move(self):
    print("I am running on 4 legs")
```

```python
dog = Dog("tonny",4,"jocky")
dog.display()
dog.move()
```
