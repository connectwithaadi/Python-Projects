```python
class Student:
  # Attributes
  name = ""
  age = 0
  gender = ""

  #Methods
  def set_info(self, name:str, age:int, gender:str):
    self.name = name
    self.age = age
    self.gender = gender

  def display(self):
    print(f"self = {self}")
    print(f"My name is {self.name}, age is {self.age} and gender is {self.gender}")
```

```python
s1 = Student()
s1.set_info("aadi",21,"male")
s1.display()
```

```python
s2 = Student()
s2.display()
```

```python
from re import I
class Student:

  #Methods
  def __init__(self, name , age , gender) -> None:
    #Attributes
    self.name = name
    self.age = age
    self.gender = gender

  def display(self) -> None:
    print(f"My name is {self.name}, age is {self.age} and gender is {self.gender}")

  def get_age(self) -> int:
    return self.age
```

```python
s1 = Student("aadi", 21, "male")
s1.display()
print(s1.get_age())
```
