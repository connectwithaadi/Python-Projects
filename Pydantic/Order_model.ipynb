```python
from pydantic import BaseModel, Field, EmailStr

class Customer(BaseModel):
  customer_id: int
  name: str
  email: EmailStr

class Product(BaseModel):
  product_id: int
  name: str
  price: float = Field(gt=0)


class Order(BaseModel):
  order_id: int
  customer: Customer
  product: Product
  quantity: int = Field(gt=0)

  @property
  def total_amount(self):

    return self.product.price * self.quantity

customer = Customer(
    customer_id = 1,
    name= "Aadi",
    email = "aadi@info.com"
)

product = Product(
    product_id = 1902,
    name="Laptop",
    price= 61590
)

order= Order(
    order_id = 404,
    customer= customer,
    product= product,
    quantity=5
)

print("=" * 50)
print("ORDER MODEL")
print("=" * 50)

print(order)
print("\nTotal Order: ", order.total_amount)

print("\nSerialized Order")

print(order.model_dump())

print("\nOrder JSON")

print(order.model_dump_json(indent = 3))
```
